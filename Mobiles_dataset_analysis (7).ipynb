# ***`8. Questions Answered by the Analysis`***

1. **Which smartphone brands dominate different price segments?**

   - **Samsung, Xiaomi, Vivo, OPPO, and Realme** have a strong presence in the budget and mid-range segments, offering devices across a wide range of price points.


2. **What features have the strongest influence on smartphone prices?**

   - **Storage capacity and RAM** have the strongest positive influence on smartphone prices.

   - Premium processors and advanced camera systems also contribute significantly to higher pricing.


3. **How do RAM, storage, processor, battery, camera, and display specifications affect pricing?**

   - Smartphones with **higher RAM and storage configurations** tend to be priced significantly higher.

   - Devices featuring **flagship processors, high-resolution cameras, larger batteries, and premium display technologies** generally command higher prices.

4. **How do operating systems and processor brands impact market positioning?**

   - **iOS devices are positioned exclusively in the premium segment**, reflecting Apple's premium branding strategy.

5. **What is the relationship between smartphone ratings and prices?**
  
   - The analysis reveals a **weak positive relationship between ratings and prices.**

   - While premium smartphones often receive good ratings, higher prices do not necessarily guarantee significantly better customer ratings.


6. **Which brands offer premium features at lower prices?**

   - **Xiaomi, Realme, Poco, and some Samsung models** provide high RAM, large storage, capable processors, and advanced camera features at relatively affordable prices.

### ***Feature Engineering***

```python
# Groups all brands outside the top 6 most frequent entries into an "Other" category for simplified analysis.
t_b=df.brand.value_counts().sort_values(ascending=False).head(6).index.tolist()
def t_p(x):
    if x not in t_b:
        return "Other"
    else:
        return x
df["b_r"]=df["brand"].apply(t_p)
```

**Observation:**
  - Groups all brands outside the top 6 most frequent entries into an "Other" category for simplified analysis and better visualization.


#**Categorizing Smartphone Processors into Standardized Processor Families**

```python
# Extracts known processor keywords from the 'Processor' column into standardized categories, defaulting unmatched entries to 'Other'.
df[df["brand"] == "Apple"][["brand", "Processor"]].value_counts()
df.loc[df["brand"] == "Apple", "Processor_Category"] = "bionic"
l=['bionic','dimensity','exynos','helio','snapdragon','unisoc','google tensor','kirin']
# 1. Standardize text to lowercase once for accurate matching
proc_lower = df["Processor"].astype(str).str.lower()

# 2. Define the search conditions for each brand
conditions = [ proc_lower.str.contains(i) for i in l]

# 3. Define the clean category names matching the order above
choices =['bionic','dimensity','exynos','helio','snapdragon','unisoc','google tensor','kirin']
# 4. Create the new column and group everything else as 'Other'
df["Processor_Category"] = np.select(conditions, choices, default="Other")
```

```python
# Merging all apple processors into bionic
df.loc[df["b_r"] == "Apple", "Processor_Category"] = "bionic"
```

```python
# Function to standardize os_version to 'vX.Y' format
def standardize_os_version(version_str):
    parts = str(version_str).split('.')
    if len(parts) > 2:
        return '.'.join(parts[:2])
    return version_str

df['os_version'] = df['os_version'].apply(standardize_os_version)
```

### Observation
- The `os_version` column has been successfully standardized to contain only up to two version numbers. For instance, `v5.1.1` is now `v5.1`, while `v12` and `v10.0` remain unchanged. This modified a cleaner and more consistent `os_version` column for further analysis.

#**Transforming Smartphone Specifications into Analysis-Friendly Categories**

```python
import pandas as pd

df['fast_charging_power'] = df['fast_charging_power'].astype(str).str.replace('No Fast Charging', '0W', case=False, regex=False)
# 1. Clean the columns by converting to numeric strings cleanly
for col, unit in [("battery_capacity", "mAh"), ("os_version", "v"), ("fast_charging_power", "W")]:
    df[col] = df[col].astype(str).str.replace(unit, "", case=False, regex=False)
l = [
    "price","RAM Size GB", "Storage Size GB", "battery_capacity", "fast_charging_power", "os_version",
    "Screen_Size", "primary_rear_camera_mp", "primary_front_camera_mp",
    "Clock_Speed_GHz", "Resolution_Width", "Resolution_Height", "rating"
]

for i in l:
    df[i] = pd.to_numeric(df[i], errors="coerce")

# 2. FIXED: Bins and matching labels list lengths
all_bins = [
    [0, 10000, 25000, 50000, 100000, 650000], # price
    [0, 1, 4, 8, 18],                     # RAM
    [0, 1, 32, 256, 1024],                # Storage
    [800, 3000, 5000, 10000, 22000],     # battery capacity
    [0, 50, 80, 240],                     # Fast Charging
    [1, 5, 10, 18],                       # Os version
    [1, 4, 6, 9],                         # Screen_Size
    [2, 5, 14, 40, 60, 200],              # primary_rear_camera_mp
    [1, 8, 16, 40, 60],                   # primary_front_camera_mp
    [0, 2, 3, 4],                         # Clock_Speed_GHz
    [120, 640, 1080, 1440, 2460],         # Resolution_Width
    [160, 640, 1480, 2160, 3840],         # Resolution_Height
    [60, 70, 80, 90, 100],                # rating
]

all_labels = [
    ['Budget','Mid-Range','Premium','Flagship','Luxury'],
    ["0-1 GB", "1-4 GB", "4-8 GB", "8-18 GB"],
    ["0-1 GB", "1-32 GB", "32-256 GB", "256-1024 GB"],
    ["800-3000 mAh", "3000-5000 mAh", "5000-10000 mAh", "10000-22000 mAh"],
    ["0-50W", "50-80W", "80-240W"],
    ["v1-v5", "v5-v10", "v10-v16"],
    ["1-4 inches", "4-6 inches", "6-9 inches"],
    ["2-5 mp", "5-14 mp", "14-40 mp", "40-60 mp", "60-200 mp"],
    ["1-8 mp", "8-16 mp", "16-40 mp", "40-60 mp"],
    ["0-2 GHz", "2-3 GHz", "3-4 GHz"],
    ["120-640 Width", "640-1080 Width", "1080-1440 Width", "1440-2460 Width"],
    ["160-640 Height", "640-1480 Height", "1480-2160 Height", "2160-3840 Height"],
    ["60-70 rating", "70-80 rating", "80-90 rating", "90-100 rating"],
]

# 3. Create categorized columns safely
for col, b, lab in zip(l, all_bins, all_labels):
    df[col + "_Category"] = pd.cut(
        df[col], bins=b, labels=lab, include_lowest=True
    )
```
